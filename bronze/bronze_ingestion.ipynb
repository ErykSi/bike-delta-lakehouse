## Celem notebooka jest załadowanie 6 plików .csv do Schema bronze jako delta tables.
## 

```python
%sql
SELECT current_catalog(), current_schema();
```

```python
%sql
USE CATALOG bike_lakehouse;
USE SCHEMA bronze;
```

```python
%python
df_cust1 = spark.read.format('csv')\
    .option('header','true')\
    .option('inferSchema','true')\
    .load('/Volumes/bike_lakehouse/bronze/raw_sources/CUST_AZ12.csv')

df_cust1.limit(5).display(5)
```

```python
df_cust1.write.mode('overwrite').saveAsTable('crm_cust')
```

```python
df_loc1 = spark.read.format('csv')\
    .option('header','true')\
    .option('inferSchema','true')\
    .load('/Volumes/bike_lakehouse/bronze/raw_sources/LOC_A101.csv')

df_loc1.limit(5).display()
df_loc1.write.mode('overwrite').saveAsTable('crm_loc')
```

```python
df_px_cat = spark.read.format('csv')\
    .option('header','true')\
    .option('inferSchema','true')\
    .load('/Volumes/bike_lakehouse/bronze/raw_sources/PX_CAT_G1V2.csv')

df_px_cat.limit(5).display()
df_px_cat.write.mode('overwrite').saveAsTable('px_cat')
```
